```python
def create_file(filename):
    with open(filename, 'x') as f:
        print(f"Success: '{filename}' has been created.")
def read_file(filename):
    with open(filename, 'r') as f:
        print(f"\n--- Everton 3 vs 3 Manchester City ---")
        print(f.read())
def append_to_file(filename):
    content = input("Enter the text you want to add: ")
    with open(filename, 'a') as f:
        f.write(content + "\n")
        print("Text successfully added.")
def main_menu():
    while True:
        print("\n--- File Management System ---")
        print("1. Create a file")
        print("2. Read a file")
        print("3. Append to a file")
        print("4. Close")

        choice = input("Choose an option (1-4): ")

        if choice == '4':
            print("Exiting program. Terima Kasih!")
            break

        if choice in ['1', '2', '3']:
            name = input("Enter the filename (e.g., test.txt): ")

            if choice == '1':
                create_file(name)
            elif choice == '2':
                read_file(name)
            elif choice == '3':
                append_to_file(name)
        else:
            print("Invalid choice. Please enter a number between 1 and 4.")

if __name__ == "__main__":
    main_menu()
```
